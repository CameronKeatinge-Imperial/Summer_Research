# Clustering agreement: ARI and NMI

Reads a whitespace-separated file with three columns (id, ground truth, predicted labels) and reports agreement between columns 2 and 3.

```python
import pandas as pd
from sklearn.metrics import adjusted_rand_score, normalized_mutual_info_score
from pathlib import Path
```

```python
CURVATURE = "OR_MMOT"
FILE = + "_clustering.txt"
CLUSTER = "ID_A"
BASE = Path(r"MY_PATH\clustering_ML\results\synthetic_data")
PATH = BASE / CLUSTER / FILE
df = pd.read_csv(PATH, sep=r"\s+", header=None, names=["id", "truth", "pred"], dtype=str)
df.head()
```

```python
truth, pred = df["truth"], df["pred"]

print(f"n = {len(df)}")
print(f"ARI = {adjusted_rand_score(truth, pred):.4f}")
print(f"NMI = {normalized_mutual_info_score(truth, pred):.4f}")
```
