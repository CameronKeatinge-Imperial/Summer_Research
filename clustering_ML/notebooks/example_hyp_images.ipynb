# Example hypernetwork graphs

```python
import hypernetx as hnx
import matplotlib.pyplot as plt

edges = {
    "$e_1$": ("$v_0$", "$w_0$","$x_0$","$y_0$"),
    "$e_2$": ("$v_0$", "$v_1$"),
    "$e_3$": ("$v_0$", "$v_2$"),
    "$e_4$": ("$w_0$", "$w_1$"),
    "$e_5$": ("$w_0$", "$v_1$"),
    "$e_6$": ("$x_0$", "$x_1$", "$x_2$"),
    "$e_8$": ("$y_0$", "$y_1$"),
}

pos = {
    "$v_0$": (-0.3,  0.3),
    "$w_0$":   (0.3, 0.3),
    "$x_0$":   (0.3, -0.3),
    "$y_0$":   (-0.3, -0.3),
    "$v_1$":   (0,  1),
    "$v_2$":   (-1,  0.6),
    "$x_1$":   (0.6, - 1),
    "$x_2$":   (1,  -0.6),
    "$w_1$":   (1,  0.6),
    "$y_1$":   (-0.75,  -0.75),
}
H = hnx.Hypergraph(edges)

highlight = "$e_1$"
colors = ["crimson" if e == highlight else "grey" for e in H.edges]

hnx.draw(
    H,
    pos=pos,
    with_edge_labels=False,
    edges_kwargs={"edgecolors": colors, "linewidths": 2},
)
plt.show()
```
